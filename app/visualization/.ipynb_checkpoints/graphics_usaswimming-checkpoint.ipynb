<h1 style="color:rgb(242, 176, 235);">Data validation</h1>

<h5 style="color:rgb(245, 72, 227);">Meet</h5>

```python
unique_meets = set()
total_meets = 0
invalid_meets = 0

for file in parquet_files:
    try:
        meet_col = pd.read_parquet(file, columns=["Meet"], engine="pyarrow")["Meet"]
    except ValueError:
        continue
    invalid_mask = meet_col.isna() | (meet_col.astype(str).str.strip() == "")

    invalid_meets += int(invalid_mask.sum())

    meet_names = meet_col[~invalid_mask].astype(str).str.strip()
    total_meets += len(meet_names)
    unique_meets.update(meet_names.unique())

print("Nombre total de meet valides :", total_meets)
print("Nombre total de meet distincts :", len(unique_meets))
print("Nombre de meet None ou vides :", invalid_meets)

print("")
```

<h5 style="color:rgb(245, 72, 227);">AgeGroup</h5>

```python
from collections import Counter

agegroup_counts = Counter()

for file in parquet_files:
    try:
        s = pd.read_parquet(file, columns=["AgeGroup"], engine="pyarrow")["AgeGroup"]
    except Exception:
        continue
    s_clean = s.fillna("").astype(str).str.strip()
    s_clean = s_clean[s_clean != ""]
    agegroup_counts.update(s_clean.value_counts().to_dict())

ordered = sorted(agegroup_counts.keys())
print("Nombre d'AgeGroup distincts :", len(ordered))
print("Performances avec AgeGroup non vide :", sum(agegroup_counts.values()))
print()
for ag in ordered:
    print(f"{ag}: {agegroup_counts[ag]:,}")
```

<h5 style="color:rgb(245, 72, 227);">SwimTime</h5>

```python
parquet_files = sorted(CLEANED_DIR.glob("*.parquet"))

allowed_pattern = r"^[0-9:.]+$"

total_null_swimtime = 0
total_invalid_format = 0

files_with_null = []
files_with_invalid = []
files_missing_column = []

for file in parquet_files:
    df = pd.read_parquet(file)

    if "SwimTime" not in df.columns:
        files_missing_column.append(str(file))
        continue

    s = df["SwimTime"]

    null_mask = s.isna()
    null_count = int(null_mask.sum())
    total_null_swimtime += null_count
    if null_count > 0:
        files_with_null.append((str(file), null_count))

    non_null = s[~null_mask].astype(str).str.strip()
    invalid_mask = ~non_null.str.match(allowed_pattern, na=False)
    invalid_count = int(invalid_mask.sum())
    total_invalid_format += invalid_count
    if invalid_count > 0:
        files_with_invalid.append((str(file), invalid_count))

print("Nombre total de SwimTime nulles     :", total_null_swimtime)
print("Nombre total de formats invalides   :", total_invalid_format)
print("Fichiers sans colonne SwimTime      :", len(files_missing_column))

if total_invalid_format == 0:
    print("OK: Toutes les valeurs SwimTime non nulles contiennent uniquement [0-9:.].")
else:
    print("ATTENTION: formats invalides detectes.")

if files_with_null:
    print("\nTop 10 fichiers avec null SwimTime:")
    for fp, c in files_with_null[:10]:
        print(f"- {fp}: {c}")

if files_with_invalid:
    print("\nTop 10 fichiers avec format SwimTime invalide:")
    for fp, c in files_with_invalid[:10]:
        print(f"- {fp}: {c}")
```

```python
parquet_files = sorted(CLEANED_DIR.glob("*.parquet"))

unique_swimmers = set()
total_swimmer_rows = 0
files_without_name = []

for file in parquet_files:
    df = pd.read_parquet(file)

    if "Name" not in df.columns:
        files_without_name.append(str(file))
        continue

    swimmer_names = df["Name"].dropna().astype(str).str.strip()
    swimmer_names = swimmer_names[swimmer_names != ""]

    total_swimmer_rows += len(swimmer_names)
    unique_swimmers.update(swimmer_names.tolist())

print("Nombre total de lignes swimmer (non vides) :", total_swimmer_rows)
print("Nombre total de swimmers uniques           :", len(unique_swimmers))
print("Fichiers sans colonne Name                 :", len(files_without_name))
```

<h5 style="color:rgb(245, 72, 227);">Rank</h5>

```python
import pandas as pd
import pyarrow.parquet as pq

total_rank_null = 0
total_rank_non_null_entier = 0
total_rank_non_valable = 0

for file in parquet_files:
    # Evite de charger tout le fichier si la colonne Rank n'existe pas
    try:
        schema = pq.read_schema(file)
    except Exception:
        continue

    if "Rank" not in schema.names:
        continue

    # Charge seulement la colonne Rank
    rank = pd.read_parquet(file, columns=["Rank"])["Rank"]

    null_mask = rank.isna()
    non_null_mask = ~null_mask

    total_rank_null += int(null_mask.sum())

    rank_str = rank[non_null_mask].astype(str).str.strip()
    entier_mask = rank_str.str.fullmatch(r"\d+", na=False)

    nb_entier = int(entier_mask.sum())
    nb_non_null = int(non_null_mask.sum())
    nb_non_valable = nb_non_null - nb_entier

    total_rank_non_null_entier += nb_entier
    total_rank_non_valable += nb_non_valable

print("Nombre de performances avec Rank null                :", total_rank_null)
print("Nombre de performances avec Rank non null et entier  :", total_rank_non_null_entier)
print("Nombre de performances avec Rank non valable         :", total_rank_non_valable)
```
